# Data exploration

This notebook explores the raw datasets.
The goal is to understand data structure, data quality issues, and inconsistencies
that informed the design of the ETL pipeline.


```python
import pandas as pd
from pathlib import Path
```

```python
BASE_DIR = Path.cwd().parent
DATA_DIR = BASE_DIR / "data"
RAW_DATA_DIR = DATA_DIR / "raw"

APPOINTMENTS_PATH = RAW_DATA_DIR / "Appointments excel.xlsx"
DOCTORS_PATH = RAW_DATA_DIR / "Doctors excel.xlsx"

df_appointments = pd.read_excel(APPOINTMENTS_PATH)
df_doctors = pd.read_excel(DOCTORS_PATH)
```
