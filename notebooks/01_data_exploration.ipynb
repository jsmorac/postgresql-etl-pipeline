# Data exploration – VIP Medical Group

This notebook explores the raw datasets provided for the technical assessment.
The goal is to understand data structure, data quality issues, and inconsistencies
that informed the design of the ETL pipeline.


```python
import pandas as pd
from pathlib import Path
```

```python
BASE_DIR = Path.cwd()
DATA_DIR = BASE_DIR / "data"
RAW_DATA_DIR = DATA_DIR / "raw"

APPOINTMENTS_PATH = RAW_DATA_DIR / "Data Engineer's Appointments Excel - VIP Medical Group.xlsx"
DOCTORS_PATH = RAW_DATA_DIR / "Data Enginner's Doctors Excel - VIP Medical Group.xlsx"

df_appointments = pd.read_excel(APPOINTMENTS_PATH)
df_doctors = pd.read_excel(DOCTORS_PATH)
```

## Key Findings

- `booking_id` contains non-numeric values (e.g., '37X')
- Appointment status values are inconsistent in casing and spelling
- A doctor_id (105) exists in appointments but not in the doctors dataset
- A significant number of booking_date values are invalid or inconsistent
- No duplicated rows were found in either dataset
